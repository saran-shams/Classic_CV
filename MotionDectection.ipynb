```python
# Template: box around the pedestrian in red shirt (x, y, width, height)
TX, TY, TW, TH = 160, 530, 110, 200
R = 20
def calculate_matching_errors(img1, img2, x, y, w, h, R):
    T = img1[y:y + h, x:x + w].astype(np.float32)

    ssd = np.full((2 * R + 1, 2 * R + 1), np.nan, np.float32)
    sad = np.full_like(ssd, np.nan)
    for dy in range(-R, R + 1):
        for dx in range(-R, R + 1):
            xx, yy = x + dx, y + dy

            if xx < 0 or yy < 0 or xx + w > img2.shape[1] or yy + h > img2.shape[0]:
                continue
            P = img2[yy:yy + h, xx:xx + w].astype(np.float32)
            e = P - T
            ssd[dy + R, dx + R] = np.mean(e ** 2) # SSD / area
            sad[dy + R, dx + R] = np.mean(np.abs(e)) # SAD / area
    return ssd, sad
def best(err):
    iy, ix = np.unravel_index(np.nanargmin(err), err.shape)

    return int(ix - R), int(iy - R)
ssd_map, sad_map = calculate_matching_errors(gray1, gray2, TX, TY, TW, TH, R)
u_ssd, u_sad = best(ssd_map), best(sad_map)
print("SSD best translation (dx, dy):", u_ssd, f" minimum mean SSD: {np.nanmin(ssd_map):.1f}")
print("SAD best translation (dx, dy):", u_sad, f" minimum mean SAD: {np.nanmin(sad_map):.1f}")
print(f"Error with no motion (0, 0): SSD = {ssd_map[R, R]:.1f} SAD = {sad_map[R, R]:.1f}")
fig, ax = plt.subplots(1, 3, figsize=(17, 5))
ax[0].imshow(frame1)
ax[0].add_patch(plt.Rectangle((TX, TY), TW, TH, fill=False, ec="red", lw=2))
ax[0].set_title("Template (red box) in frame t"); ax[0].axis("off")
for a, m, name, b in [(ax[1], ssd_map, "SSD", u_ssd), (ax[2], sad_map, "SAD", u_sad)]:
    im = a.imshow(m, origin="lower", extent=(-R - .5, R + .5, -R - .5, R + .5))
    a.plot(*b, "r*", ms=15, label=f"minimum {b}")
    a.set_xlabel("horizontal displacement (pixels)"); a.set_ylabel("vertical displacement (pixels)")
    a.set_title(f"{name} matching error"); a.legend(); plt.colorbar(im, ax=a, fraction=0.046)
plt.tight_layout(); plt.show()
```

```python
def image_derivatives(I1, I2):
    # Convert frames to float32 to support negative and fractional values.
    I1 = I1.astype(np.float32)
    I2 = I2.astype(np.float32)
    Ix = 0.5 * (
        cv2.Sobel(I1, cv2.CV_32F, 1, 0, ksize=3) +
        cv2.Sobel(I2, cv2.CV_32F, 1, 0, ksize=3)
    ) / 8.0
    Iy = 0.5 * (
        cv2.Sobel(I1, cv2.CV_32F, 0, 1, ksize=3) +
        cv2.Sobel(I2, cv2.CV_32F, 0, 1, ksize=3)
    ) / 8.0
    It = I2 - I1
    return Ix, Iy, It

Ix, Iy, It = image_derivatives(gray1, gray2)
plt.figure(figsize=(16, 4.5))

for i, (img, name) in enumerate([
    (Ix, "Ix (spatial derivative, x)"),
    (Iy, "Iy (spatial derivative, y)"),
    (It, "It (frame differencing)")
]):
    lim = np.percentile(np.abs(img), 99.5)
    plt.subplot(1, 3, i + 1)
    plt.imshow(img, cmap="gray", vmin=-lim, vmax=lim)
    plt.title(name)
    plt.axis("off")
plt.tight_layout()
plt.show()

difference = cv2.absdiff(gray1, gray2)
_, motion_mask = cv2.threshold(
    difference, 20, 255, cv2.THRESH_BINARY
)
motion_mask_clean = cv2.morphologyEx(
    motion_mask,
    cv2.MORPH_OPEN,
    np.ones((3, 3), np.uint8)
)
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(gray1, cmap="gray")
plt.title("Previous Frame")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(difference, cmap="gray")
plt.title("Absolute Frame Difference")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(motion_mask_clean, cmap="gray")
plt.title("Thresholded Motion Mask")
plt.axis("off")
plt.tight_layout()
plt.show()

changed = np.count_nonzero(motion_mask_clean)
print("Changed pixels:", changed)
print(f"Changed pixel percentage: {100 * changed / motion_mask_clean.size:.2f}%")
```

```python
S = 4
def shrink(g):
    small = cv2.resize(g, None, fx=1 / S, fy=1 / S, interpolation=cv2.INTER_AREA)
    return cv2.GaussianBlur(small, (3, 3), 0.8).astype(np.float32)

A, B = shrink(gray1), shrink(gray2)
Ix_s, Iy_s, It_s = image_derivatives(A, B)
bx, by, bw, bh = TX // S, TY // S, TW // S, TH // S
r = 7

def window_tensor(cx, cy):
    s = (slice(cy - r, cy + r + 1), slice(cx - r, cx + r + 1))
    ix, iy = Ix_s[s].ravel(), Iy_s[s].ravel()
    return np.linalg.eigvalsh([[ix @ ix, ix @ iy], [ix @ iy, iy @ iy]])[0]

cands = [(cx, cy) for cy in range(by + r, by + bh - r) for cx in range(bx + r, bx + bw - r)]
cx, cy = max(cands, key=lambda p: window_tensor(*p))
s = (slice(cy - r, cy + r + 1), slice(cx - r, cx + r + 1))
ix, iy, it = Ix_s[s].ravel(), Iy_s[s].ravel(), It_s[s].ravel()
(u_lk, v_lk), *_ = np.linalg.lstsq(np.stack([ix, iy], 1), -it, rcond=None)
# Print the estimated flow in the downsampled image coordinates.
print(f"Least-squares flow (small image): u = {u_lk:.2f}, v = {v_lk:.2f}")
# Multiply by the shrink factor to express the estimated flow in original pixels.
# Compare the result with the SSD displacement estimate.
print(f"Scaled to full resolution: u = {u_lk * S:.1f}, v = {v_lk * S:.1f} (SSD gave {u_ssd})")

def draw_lines(ax, ix, iy, it, lim=6, n=60):
    # Define the plotting range for horizontal and vertical flow components.
    t = np.array([-lim, lim])
    # Select up to n pixels with the strongest spatial gradients.
    for k in np.argsort(-np.hypot(ix, iy))[:n]:
        # Retrieve the horizontal gradient, vertical gradient, and temporal derivative.
        a, b, d = ix[k], iy[k], it[k]
        # Plot the constraint line a*u + b*v = -d.
        # Choose the equation form that avoids dividing by the smaller coefficient.
        if abs(b) >= abs(a):
            ax.plot(t, -(d + a * t) / b, lw=1, alpha=.8)
        else:
            ax.plot(-(d + b * t) / a, t, lw=1, alpha=.8)
    # Draw reference axes at zero flow.
    ax.axhline(0, color="gray", lw=.5)
    ax.axvline(0, color="gray", lw=.5)
    # Set equal scales and limits for horizontal and vertical flow.
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect("equal")
    # Label the flow components.
    ax.set_xlabel("u")
    ax.set_ylabel("v")

# Find the window pixel with the strongest spatial gradient.
k1 = np.argmax(np.hypot(ix, iy))
# Create three plots to illustrate the optical-flow constraints.
fig, ax = plt.subplots(1, 3, figsize=(17, 5))
# Display the downsampled image and mark the selected 15 x 15 window.
ax[0].imshow(A, cmap="gray")
ax[0].add_patch(plt.Rectangle((cx - r - .5, cy - r - .5), 2 * r + 1, 2 * r + 1, fill=False, ec="red", lw=2))
ax[0].set_xlim(bx - 15, bx + bw + 15)
ax[0].set_ylim(by + bh + 15, by - 15)
ax[0].set_title("Window of neighbors (red)")
ax[0].axis("off")
# Plot the flow constraint from one pixel.
# One equation gives a line of possible (u, v) values, not a unique solution.
draw_lines(ax[1], ix[[k1]], iy[[k1]], it[[k1]])
ax[1].set_title("ONE pixel: a line of possible (u, v)")
# Plot constraint lines from all selected window pixels.
# Their approximate intersection indicates the least-squares flow estimate.
draw_lines(ax[2], ix, iy, it)
ax[2].plot(u_lk, v_lk, "k*", ms=16)
ax[2].set_title("Neighbors: lines cross at the answer (star)")
# Adjust subplot spacing and display all three plots.
plt.tight_layout()
plt.show()
```

```python
# Detect up to 300 strong corner features in the first grayscale frame.
# qualityLevel sets the minimum accepted corner strength relative to the strongest corner.
# minDistance enforces at least 8 pixels between detected features.
# blockSize specifies the neighborhood used to calculate corner quality.
features = cv2.goodFeaturesToTrack(gray1, maxCorners=300, qualityLevel=0.01, minDistance=8, blockSize=7)

# Stop execution if no trackable features were detected.
if features is None or len(features) == 0:
    raise RuntimeError("No trackable features were detected. Try another frame pair.")

# Track the detected features from the first frame to the second using pyramidal Lucas–Kanade optical flow.
# winSize defines the local search window; maxLevel=3 allows a pyramid with up to 3 additional levels.
# criteria stops the iterative search after 30 iterations or when the change falls below 0.01.
next_points, status, errors = cv2.calcOpticalFlowPyrLK(
    gray1, gray2, features, None, winSize=(21, 21), maxLevel=3,
    criteria=(cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 30, 0.01)
)

# Keep only points that Lucas–Kanade successfully tracked.
valid = status.ravel() == 1
# Extract the original positions of successfully tracked features.
good_old = features.reshape(-1, 2)[valid]
# Extract their corresponding positions in the second frame.
good_new = next_points.reshape(-1, 2)[valid]
# Calculate each tracked feature's horizontal and vertical displacement (u, v).
vectors = good_new - good_old
# Classify points with displacement greater than 1.5 pixels as moving.
moving = np.linalg.norm(vectors, axis=1) > 1.5
# Copy the second frame to draw the optical-flow results without altering the original.
output_sparse = frame2.copy()
# Set the arrow-length amplification factor for easier visualization.
GAIN = 3

# Process each tracked point, its new position, and its movement classification.
for (x1, y1), (x2, y2), mv in zip(good_old, good_new, moving):
    # If the displacement exceeds the threshold, draw an amplified green arrow.
    if mv:
        end = (int(round(x1 + GAIN * (x2 - x1))), int(round(y1 + GAIN * (y2 - y1))))
        cv2.arrowedLine(output_sparse, (int(x1), int(y1)), end, (0, 255, 0), 2, tipLength=0.3)
    # Otherwise, mark the point as approximately static with a blue circle.
    else:
        cv2.circle(output_sparse, (int(x1), int(y1)), 3, (255, 0, 0), -1)

# Display the second frame with green motion arrows and blue static-point markers.
plt.figure(figsize=(12, 7))
plt.imshow(output_sparse)
plt.title("Lucas–Kanade Sparse Optical Flow (green arrows x3 = moving points)")
plt.axis("off")
plt.show()

# Print the total number of features detected in the first frame.
print("Initially detected features:", len(features))
# Print how many detected features were successfully tracked in the second frame.
print("Successfully tracked points:", int(valid.sum()))
# Print the counts of points classified as moving and approximately static.
print("Moving points:", int(moving.sum()), " Static points:", int((~moving).sum()))
# Calculate and print the mean horizontal and vertical displacement of moving points.
print("Mean displacement of moving points (u, v):", np.round(vectors[moving].mean(0), 2))
```

```python
# Calculate dense optical flow between the previous and next grayscale frames.
flow = cv2.calcOpticalFlowFarneback(gray1, gray2, None, pyr_scale=0.5, levels=3, winsize=21,
iterations=3, poly_n=5, poly_sigma=1.2, flags=0)
# Extract horizontal and vertical motion components from the flow field.
flow_x, flow_y = flow[..., 0], flow[..., 1]
# Calculate each pixel's motion magnitude and direction angle.
magnitude, angle = cv2.cartToPolar(flow_x, flow_y)
# Create an HSV image to visualize motion direction and magnitude using color.
hsv = np.zeros((*gray1.shape, 3), np.uint8)
# Encode motion direction as hue, mapping angles from 0 to 2π into OpenCV's hue range.
hsv[..., 0] = (angle * 90 / np.pi).astype(np.uint8)
# Set saturation to maximum so motion directions appear in vivid colors.
hsv[..., 1] = 255
# Normalize motion magnitudes to the range 0–255 for brightness visualization.
hsv[..., 2] = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
# Convert the HSV motion visualization into RGB format for Matplotlib.
flow_rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)
# Create three plots showing the original frame, dense flow, and motion magnitude.
plt.figure(figsize=(15, 5))
# Display the previous video frame as a reference.
plt.subplot(1, 3, 1)
plt.imshow(frame1)
plt.title("Previous Frame")
plt.axis("off")
# Display the dense optical-flow visualization.
# Color represents motion direction, while brightness represents relative magnitude.
plt.subplot(1, 3, 2)
plt.imshow(flow_rgb)
plt.title("Dense Optical Flow")
plt.axis("off")
# Display the magnitude map, where brighter regions indicate stronger estimated motion.
plt.subplot(1, 3, 3)
plt.imshow(magnitude, cmap="inferno")
plt.title("Motion Magnitude")
plt.colorbar(fraction=0.046)
plt.axis("off")
# Adjust subplot spacing and display all three visualizations.
plt.tight_layout()
plt.show()
# Print the average estimated motion magnitude across all pixels.
print("Mean flow magnitude:", round(float(magnitude.mean()), 3))
# Print the largest estimated motion magnitude in the flow field.
print("Maximum flow magnitude:", round(float(magnitude.max()), 3))
# Create a copy of the second frame for drawing motion vectors.
vector_image = frame2.copy()
# Set the grid spacing between arrows and amplify arrow lengths for visibility.
step, arrow_scale = 12, 3.0
# Sample the dense flow field at regularly spaced pixel locations.
for y in range(step, H - step, step):
    for x in range(step, W - step, step):
        # Retrieve the horizontal and vertical flow components at this pixel.
        dx, dy = flow[y, x]
        # Draw arrows only for moderate motion magnitudes between 1.5 and 30 pixels.
        # This excludes small movements and unusually large estimated displacements.
        if 1.5 < np.hypot(dx, dy) < 30:
            # Draw a green arrow from the sampled pixel along its amplified motion vector.
            cv2.arrowedLine(vector_image, (x, y), (int(round(x + arrow_scale * dx)), int(round(y + arrow_scale * dy))),
            (0, 255, 0), 2, tipLength=0.3)
# Display the second frame overlaid with sampled dense motion vectors.
plt.figure(figsize=(12, 7))
plt.imshow(vector_image)
plt.title(f"Dense Motion Vectors (grid spacing = {step} px, arrows x{arrow_scale:g})")
plt.axis("off")
plt.show()
```

```python
# Find a window on a moving edge: strong gradient in ONE direction (lam2 >> lam1) and motion present
r = 7 # Set the radius of the circular window
k = (2 * r + 1, 2 * r + 1) # Define the window size for local summation
Sxx = cv2.boxFilter(Ix_s * Ix_s, -1, k, normalize=False) # Sum squared horizontal gradients within each window
Sxy = cv2.boxFilter(Ix_s * Iy_s, -1, k, normalize=False) # Sum products of horizontal and vertical gradients
Syy = cv2.boxFilter(Iy_s * Iy_s, -1, k, normalize=False) # Sum squared vertical gradients within each window
tr, det = Sxx + Syy, Sxx * Syy - Sxy ** 2 # Calculate the trace and determinant of the local structure tensor
disc = np.sqrt(np.maximum(tr ** 2 / 4 - det, 0)) # Calculate the term used to obtain the eigenvalues
lam1, lam2 = tr / 2 - disc, tr / 2 + disc # Calculate the smaller and larger eigenvalues

flow_small = cv2.resize(flow, (A.shape[1], A.shape[0])) / S # Resize the flow and scale it to small-image pixel units

moving_zone = cv2.boxFilter(np.hypot(flow_small[..., 0], flow_small[..., 1]), -1, k) > 1.0 # Identify windows with sufficient motion

ok = moving_zone & (lam2 > np.percentile(lam2[moving_zone], 50)) # Keep moving windows with stronger-than-median gradients

ok[:r + 2] = ok[-(r + 2):] = False; ok[:, :r + 2] = ok[:, -(r + 2):] = False # Exclude border windows that may extend outside the image
edge_score = np.where(ok, lam2 / (lam1 + 1e-3), -1) # Score valid windows by how strongly their gradients favour one direction
ey, ex = np.unravel_index(np.argmax(edge_score), edge_score.shape) # Find the window with the strongest directional gradient
s = (slice(ey - r, ey + r + 1), slice(ex - r, ex + r + 1)) # Define the selected window around the detected location
circ = ((np.mgrid[-r:r + 1, -r:r + 1] ** 2).sum(0) <= (r + .5) ** 2).ravel() # Create a circular mask for pixels inside the aperture
ix, iy, it = Ix_s[s].ravel()[circ], Iy_s[s].ravel()[circ], It_s[s].ravel()[circ] # Extract horizontal, vertical, and temporal gradients inside the circle
g2 = ix ** 2 + iy ** 2 # Calculate the squared spatial-gradient magnitude for each pixel
# Normal flow: motion along the gradient direction n, found by least squares
nvec = np.array([np.sum(ix * np.hypot(ix, iy)), np.sum(iy * np.hypot(ix, iy))]) # Estimate the dominant gradient direction
nvec /= np.linalg.norm(nvec) # Normalize the gradient direction to unit length
normal_speed = -np.sum(np.hypot(ix, iy) * it) / np.sum(g2)
# Estimate motion speed along the gradient using the optical-flow constraint
normal_flow = normal_speed * nvec # Convert the estimated speed into a 2D normal-flow vector
true_flow = flow_small[s].reshape(-1, 2).mean(0) # Calculate the average real motion vector inside the selected window
print(f"Window at (x={ex}, y={ey}) in the small image, eigenvalues: {lam1[ey, ex]:.0f} (small), {lam2[ey, ex]:.0f} (large)") # Display the selected location and its eigenvalues
print(f"Real motion (green): u = {true_flow[0]:.2f}, v = {true_flow[1]:.2f}") # Print the real horizontal and vertical motion
print(f"Normal flow (red): u = {normal_flow[0]:.2f}, v = {normal_flow[1]:.2f}") # Print the estimated normal-flow components
print(f"Part of the real motion ALONG the edge that the window cannot see: {abs(true_flow @ np.array([-nvec[1], nvec[0]])):.2f} px") # Calculate the motion component parallel to the edge, which normal flow cannot recover
pad = 22 # Set the size of the image region displayed around the selected window
y0, y1, x0, x1 = max(0, ey - pad), ey + pad, max(0, ex - pad), ex + pad # Define the display-region boundaries
yy, xx = np.mgrid[y0:y1, x0:x1] # Generate pixel coordinates for the display region
aperture = (xx - ex) ** 2 + (yy - ey) ** 2 <= (r + .5) ** 2 # Create a circular mask around the selected location
fig, ax = plt.subplots(1, 3, figsize=(17, 5.5)) # Create three side-by-side plots
for a, img, title in [(ax[0], A, "Frame t, circular aperture"), (ax[1], B, "Frame t+1")]: # Iterate over the two consecutive frames
    a.imshow(img[y0:y1, x0:x1], cmap="gray", extent=(x0 - .5, x1 - .5, y1 - .5, y0 - .5)) # Display the selected image region in grayscale
    a.add_patch(plt.Circle((ex, ey), r + .5, fill=False, ec="yellow", lw=2)) # Draw the circular aperture around the selected window
    a.set_title(title); a.axis("off") # Set the subplot title and hide the axes
G = 4 # Set the scale factor for displaying motion vectors
ax[0].arrow(ex, ey, *(true_flow * G), color="lime", width=.15, head_width=1.2, length_includes_head=True) # Draw the real motion vector in green
ax[0].arrow(ex, ey, *(normal_flow * G), color="red", width=.15, head_width=1.2, length_includes_head=True) # Draw the normal-flow vector in red
ax[0].plot([], [], color="lime", label="real motion"); ax[0].plot([], [], color="red", label="normal flow (all we can see)") # Add legend entries for both motion vectors
ax[0].legend(loc="lower left", fontsize=8) # Display the vector legend
draw_lines(ax[2], ix, iy, it, lim=6, n=80) # Plot the optical-flow constraint lines for the selected pixels
ax[2].plot(*true_flow, "g*", ms=15, label="real motion"); ax[2].plot(*normal_flow, "r*", ms=12, label="normal flow") # Mark the real and normal-flow vectors
ax[2].legend(loc="lower left", fontsize=8); ax[2].set_title("Constraint lines of the aperture (mostly one direction)") # Add the legend and describe the constraint-line plot
plt.tight_layout(); plt.show()
```

```python
def horn_schunck(I1, I2, alpha=10.0, iterations=300):
    # Horn–Schunck dense optical flow. alpha**2 = 1/lambda (smoothness weight).
    I1 = I1.astype(np.float32); I2 = I2.astype(np.float32) # Convert both frames to 32-bit floating point
    # Steps 1 and 2: spatial and temporal gradients
    Ix, Iy, It = image_derivatives(I1, I2) # Calculate horizontal, vertical, and temporal image gradients
    # Step 3: initialize the flow field
    u = np.zeros_like(I1); v = np.zeros_like(I1) # Initialize horizontal and vertical flow components to zero
    # Weighted average of the 8 neighbors (centre excluded)
    avg = np.array([[1, 2, 1], [2, 0, 2], [1, 2, 1]], np.float32) / 12.0 # Define normalized weights for averaging neighboring flow values
    denominator = alpha ** 2 + Ix ** 2 + Iy ** 2 # Precompute the denominator used in each flow update
    # Step 4: iterate
    for _ in range(iterations): # Repeatedly refine the flow estimates
        u_bar = cv2.filter2D(u, -1, avg, borderType=cv2.BORDER_REPLICATE) # Compute the locally averaged horizontal flow
        v_bar = cv2.filter2D(v, -1, avg, borderType=cv2.BORDER_REPLICATE) # Compute the locally averaged vertical flow
        correction = (Ix * u_bar + Iy * v_bar + It) / denominator # Calculate the correction from the optical-flow constraint
        u = u_bar - Ix * correction # Update the horizontal flow estimate
        v = v_bar - Iy * correction # Update the vertical flow estimate
    return u, v # Return the estimated dense horizontal and vertical flow fields

def to_full(u_small, v_small):
    # Back to full resolution: resize and multiply the flow by S
    return (cv2.resize(u_small, (W, H)) * S, cv2.resize(v_small, (W, H)) * S) # Resize both flow fields and scale their displacement values to full-resolution pixels

def flow_to_rgb(u, v, max_mag=10.0):
    mag, ang = cv2.cartToPolar(u, v, angleInDegrees=True) # Convert horizontal and vertical flow into motion magnitude and direction
    hsv = np.zeros((*u.shape, 3), np.uint8) # Create an empty HSV image for visualizing optical flow
    hsv[..., 0] = (ang / 2).astype(np.uint8) # Encode motion direction as HSV hue
    hsv[..., 1] = 255 # Set full saturation for all pixels
    hsv[..., 2] = np.clip(255 * mag / max_mag, 0, 255).astype(np.uint8) # Encode motion speed as brightness, clipping values to the valid range
    return cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB) # Convert the HSV flow visualization to RGB

u_hs, v_hs = horn_schunck(A, B, alpha=10.0, iterations=300) # Estimate dense optical flow between the two frames
u_full, v_full = to_full(u_hs, v_hs) # Convert the flow fields to full-image resolution
vec_hs = frame2.copy() # Copy the second frame to draw motion vectors on it
step = 12 # Set the spacing between sampled motion vectors
for y in range(step, H - step, step): # Iterate over image rows at the selected spacing
    for x in range(step, W - step, step): # Iterate over image columns at the selected spacing
        du, dv = u_full[y, x], v_full[y, x] # Retrieve horizontal and vertical flow at the current pixel
        if 1.5 < np.hypot(du, dv) < 30: # Display only motion vectors with magnitudes between 1.5 and 30 pixels
            cv2.arrowedLine(vec_hs, (x, y), (int(round(x + 3 * du)), int(round(y + 3 * dv))), (0, 255, 0), 2, tipLength=0.3) # Draw a green arrow showing the motion direction and scaled displacement

fig, axes = plt.subplots(1, 3, figsize=(18, 5.5)) # Create three side-by-side plots
axes[0].imshow(gray1, cmap="gray"); axes[0].set_title("First Frame") # Display the first grayscale frame
axes[1].imshow(flow_to_rgb(u_full, v_full)); axes[1].set_title("Horn–Schunck Dense Optical Flow") # Display the dense flow using color-coded direction and magnitude
axes[2].imshow(vec_hs); axes[2].set_title("Horn–Schunck Motion Vectors (arrows x3)") # Display the second frame with scaled motion arrows
for a in axes: a.axis("off")
plt.tight_layout(); plt.show()

# Effect of the smoothness weight alpha
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5)) # Create three plots to compare different smoothness weights
for a, al in zip(axes, [1.0, 10.0, 50.0]): # Iterate over the three alpha values
    uu, vv = horn_schunck(A, B, alpha=al, iterations=300) # Estimate optical flow using the current smoothness weight
    a.imshow(flow_to_rgb(*to_full(uu, vv))); a.set_title(f"alpha = {al:g}"); a.axis("off")
plt.tight_layout(); plt.show()
```
