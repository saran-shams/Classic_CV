```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# LAB TASK 1:
def run_lab_task_1(image_path):
    print("--- Running Lab Task 1 ---")
    img = cv2.imread('images.jpeg', cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(f"Could not load image at {image_path}")

    laplacian_direct = cv2.Laplacian(img, cv2.CV_64F)
    laplacian_direct_abs = cv2.convertScaleAbs(laplacian_direct)

    blurred_img = cv2.GaussianBlur(img, (5, 5), sigmaX=1.5, sigmaY=1.5)
    laplacian_log = cv2.Laplacian(blurred_img, cv2.CV_64F)
    laplacian_log_abs = cv2.convertScaleAbs(laplacian_log)

    plt.figure(figsize=(12, 6))

    plt.subplot(1, 2, 1)
    plt.imshow(laplacian_direct_abs, cmap='gray')
    plt.title("Original Laplacian (No Blur)")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(laplacian_log_abs, cmap='gray')
    plt.title("Blurred-then-Laplacian (LoG)")
    plt.axis('off')

    plt.tight_layout()
    plt.show()


# LAB TASK 2:
def run_lab_task_2(image_path):
    print("--- Running Lab Task 2 ---")
    img = cv2.imread('images.jpeg', cv2.IMREAD_GRAYSCALE)
    if img is None:
        return

    threshold_pairs = [(50, 100), (100, 200), (150, 300)]
    edge_maps = []

    for low, high in threshold_pairs:
        edges = cv2.Canny(img, low, high)
        edge_maps.append(edges)

    plt.figure(figsize=(15, 5))

    for i, (low, high) in enumerate(threshold_pairs):
        plt.subplot(1, 3, i + 1)
        plt.imshow(edge_maps[i], cmap='gray')
        plt.title(f"Canny Thresholds: ({low}, {high})")
        plt.axis('off')

    plt.tight_layout()
    plt.show()



if __name__ == "__main__":
    my_image_path = "input_image.jpg"

    run_lab_task_1(my_image_path)
    run_lab_task_2(my_image_path)
```
